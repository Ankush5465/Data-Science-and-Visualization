```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    'Date': [
        '2026-01-01','2026-01-02','2026-01-03',
        '2026-01-04','2026-01-05','2026-01-07',
        '2026-01-08','2026-01-09','2026-01-10',
        '2026-01-11','2026-01-12','2026-01-14',
        '2026-01-15','2026-01-16','2026-01-17',
        '2026-01-18','2026-01-19','2026-01-21',
        '2026-01-22','2026-01-23','2026-01-24',
        '2026-01-25','2026-01-26','2026-01-28',
        '2026-01-29'
    ],

    'Sales': [
        12000,15000,np.nan,
        18000,20000,17000,
        27000,np.nan,25000,
        28000,30000,26000,
        32000,35000,np.nan,
        33000,36000,38000,
        40000,42000,39000,
        45000,np.nan,48000,
        50000
    ],

    'Customers':[
        40,45,35,50,55,
        48,60,58,np.nan,
        70,75,68,80,85,
        82,78,90,92,
        95,100,96,105,
        110,115,np.nan
    ]
}
#create Dataframe
df = pd.DataFrame(data)

# Display the dataframe
print(df)
```

```python
# checking the data tpye
print(df.dtypes)

# convert to Datetime format for date column
df['Date']=pd.to_datetime(df['Date'])

print("\n After Type conversion ")
print(df.dtypes)
```

```python
#create the index for data column
df.set_index('Date',inplace=True)

#Display the Data frame
print(df.head(7))
```

```python
# Generate the Data range  - Eg Daily Date RAnge
Complete_dates=pd.date_range(start=df.index.min(),
                             end=df.index.max(),
                             freq='D') # freq=D|min|s|h|ME|QE|YE
print(Complete_dates)
```

```python
# Extract Year,Month,Day,Weekday
date_info=pd.DataFrame(index=df.index)

#Extract date component
date_info['Year']=df.index.year
date_info['Month']=df.index.month
date_info['Day']=df.index.all
date_info['Weekday']=df.index.day_name()

#display the result
print(date_info.head(10))
```

```python
# find the Missing date and missing value
missing_data={
    'Date':pd.date_range(start='2026-01-01',
                        periods=7,
                        freq='D'),
    'Sales':[12000,15000,np.nan,18000,np.nan,22000,25000]
}

df_missing=pd.DataFrame(missing_data)
df_missing.set_index('Date',inplace=True)
print(df_missing)
```

```python
# find the missing value
print(df_missing.isnull())
```

```python
print(df_missing.isnull().sum())
```

```python
#fill the Missing value
#1. backwardfill
#2. Forwardfill
#3. interpolation
```

```python
df_ffill=df_missing.ffill()
print(df_ffill)
```

```python
df_backfill=df_missing.bfill()
print(df_backfill)
```

```python
df_interpolated=df_missing.interpolate(method='linear')
print(df_interpolated)
```

```python
#Resampling
#1) upsampling  => (lowerlevel to higher level) Daily -> week -> month -> QE -> HE -> YE
#2) downsampling => (Higher level to lower level)  YE -> HE -> ME -> week -> day -> Hours -> Min -> Seconds
```

```python
# perform weekly downsampling

#weekly Averages Sales
weekly_avg=df['Sales'].resample('W').mean()

# weekly minimun Sales
weekly_min=df['Sales'].resample('W').min()

#weekly maximum sales
weekly_max=df['Sales'].resample('W').max()

print("\n Average sales for week wise")
print(weekly_avg)
print('\n Minimum sales:',weekly_min)
print('\n Maximum sales:',weekly_max)
```
