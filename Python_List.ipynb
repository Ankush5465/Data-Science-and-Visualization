```python
# extend function
sec_a =["Chirag","Devi","Jhansi"]
sec_b =["Kiran","Ganesh","Lakshman"]
sec_a.extend(sec_b)
print(sec_a)
print(sec_b)
```

```python
# remove an item in the list
sec_a_students = ["Chirag","jhansi","Devi","Kalinga","Manish"]
sec_a_students.remove("Devi")
print(sec_a_students)
del sec_a_students[1]
print(sec_a_students)
sec_a_students.pop()
print(sec_a_students)
```

```python
# List Slicing
subjects =["python",
           "Statics",
           "SQL",
           "Java",
           "OS",
           "Machine Learning"]
print(subjects[1:4])
print(subjects[:3])
print(subjects[4:])
print(subjects[::2])

#length of the list
print("Length of the list: ",len(subjects))
```
