#Tuple
## ordered collection of items
## symbol ()
## Tuple Immutable
## supports index


```python
# create a tuple
students = (101,"chirag","CSE",8.9)
print(students)

# access by index
print("Student Department:",students[2])

# Access by negative index
print("Accessing by negative index:",students[-2])
```

```python
subjects = ("DSV","Machine Learning","OS","cloud","SQL","Statistics","Python")
print(subjects)

# find the length of the tuple
print("Length of the tuple:",len(subjects))
```

```python
# tuple slicing
print(subjects[3:])
print(subjects[1:5])
```

```python
# tuple is mutable or immutable
print(students)
students[1]="Manish"
print(students)
```

```python
# Nested tuple
student_info=(1005,"Rahul",(89,98,78,75))
print(student_info)
print("Nested tuple displaying 3rd subject mark:",student_info[2][2])
```

```python
position = student_info.index("Rahul")
print("Rahul data present in the index:",position)
```

# SET
## unordered collection of items
## mutable
## symbol {}
## NO duplication
## it will give the unique values
## union , intersection, difference, symmetric difference, superset

```python
# create a set
subjects_info={"Data Science and visualization",
               "SQL",
               "Statistics",
               "Mschine Learning"}
print(subjects_info)
```

```python
studentsname = {"Amit","Sanjiv","Kalai","Amit","Kiran","Neha","sanjiv"}
print(studentsname)
```

```python
print(studentsname[2])
```

```python
# adding an element in the set
studentsname.add("Priya")
print(studentsname)

# update the set
studentsname.update(["Anil","Nandani","Balaji"])
print(studentsname)
```

```python
# remove an item in the set
# studentsname.remove("Amit")
# print(studentsname)

# remove an item
# studentsname.remove("Amit")
# print(studentsname)

# students
studentsname.discard("Priya")
print(studentsname)

studentsname.discard("Amit")
print(studentsname)
```

```python
# clear an delete
numbers = {89,78,68,97,69}
numbers.clear()
print(numbers)

del numbers
print(numbers)
```

```python
# find the element in the set
print(studentsname)


# find the element Anil
print("Anil" in studentsname)
```

```python
# set operation
# union
# Intersection
# difference
# symmetric difference

python_students={"rahul","Priya","Hari","Ganesh"}
java_students={"rahul","Hari","Kiran","Ravi","Giri"}

res_union=python_students.union(java_students)
print(res_union)
```

```python
res_inter=python_students.intersection(java_students)
print(res_inter)
```

```python
res_diff=python_students.difference(java_students)
print(res_diff)
```

```python
lst=[101,102,103,104,101,106,102,103]
uniqu_rollno=set(lst)
print(uniqu_rollno)
unique_1st_rollno = list(uniqu_rollno)
print(unique_1st_rollno)
```
