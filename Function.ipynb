Variable length Argument(*args)

```python
def add(*num):
  total =0
  for X in num:
    total +=X
  return (total)
res1=add(10,20)
print(res1)
res2=add(1,20,30,40,50)
print(res2)
```

```python
import numpy as np
n = int(input("How many Numbers"))
numbers = []
for i in range(n):
  X = int(input("Enter a number"))
  numbers.append(X)

arr = np.array(numbers)
print("Shape",arr.shape)
print("Dim",arr.ndim)
print("type",type(arr))
print("Array type",arr.dtype)
```

```python
def create_Array(n):
  numbers=[]
  for i in range(n):
    X = int(input("Enter a number"))
    numbers.append(X)
  return np.array(numbers)
n=int(input("How many numbers"))
arr = create_Array(n)
print("Shape",arr.shape)
print("Dim",arr.ndim)
print("type",type(arr))
print("Array type",arr.dtype)

```

```python
def cal(a,b):
  return a+b, a-b, a*b, a/b
add,sub,mul,div = cal(10,20)
print(add,sub,mul,div)
```

```python
# Input number of students
n = int(input("Enter number of students: "))

marks = []

# Input marks of each student
for i in range(n):
    m = float(input(f"Enter marks of student {i + 1}: "))
    marks.append(m)

# Calculate minimum, maximum and average
minimum = min(marks)
maximum = max(marks)
average = sum(marks) / n

# Print results
print("Minimum marks:", minimum)
print("Maximum marks:", maximum)
print("Average marks:", average)
```

# Reshaping

```python
data = np.arange(1,13)
a = data.reshape(3,4)
b = data.reshape(4,3)
c = data.reshape(2,-1)
print(f'1-D data\n{data}')
print(f'2-D 3X4 data\n{a}')
print(f'2-D 4X3 data\n{b}')
print(f'2-D 2Xinfar data\n{c}')
```

# Flatten() and ravel()

```python
data = np.array([[1,2,3],[4,5,6]])
print("Orig data",data.ndim)
flat1 = data.flatten()
print("Flatten flat1",flat1)
flat2 = data.ravel()
print("ravel flat2",flat2)
```
